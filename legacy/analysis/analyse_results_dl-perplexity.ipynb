```python
import pandas as pd
```

```python
df = pd.read_csv('./result_qpp_union/perplexity/mt5_3.csv')
df = df[df.with_0shot_posterior==False]
```

```python
x = df[(df.alpha==1)&(df.qpp_method=='nqc')][['r', 'rho', 'tau']].values[0]
print(round(x[0],4), '\t', round(x[1],4), '\t', round(x[2],4))
```

```python
check_qpp = 'nqc'

max_r = df[(df.qpp_method==check_qpp)&(df.alpha<1)&(df.alpha>0)].r.max()
x = df[(df.qpp_method==check_qpp)&(df.r==max_r)][['r', 'rho', 'tau']].values[0]
print(round(x[0],4), '\t', round(x[1],4), '\t', round(x[2],4))
```

```python
x = df[(df.alpha==0)&(df.qpp_method==check_qpp)][['r', 'rho', 'tau']].values[0]
print(round(x[0],4), '\t', round(x[1],4), '\t', round(x[2],4))
```
