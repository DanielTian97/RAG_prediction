```python
import json
import numpy as np
import math

answer_path = '../../rag_utility/gen_results'
```

```python
f = open(f'{answer_path}/random_answers_0shot_5calls_0_0_bm25_dl_19_prompt1.json')
ans = json.load(f)
f.close()

f = open(f'{answer_path}/random_answers_0shot_5calls_0_0_bm25_dl_20_prompt1.json')
ans.update(json.load(f))
f.close()
```

```python
posterior_dict = {}

for qid in ans:
    answers = ans[qid]['0']
    mean_probs = []
    for a in answers.values():
        mean_probs.append(np.mean(eval(a['probs'])))
    posterior_dict.update({str(qid): np.mean(mean_probs)})
```

```python
import pickle

with open('./res/dl_mean_posteriors.pkl', 'wb') as f:
    pickle.dump(posterior_dict, f)
    f.close()
```
