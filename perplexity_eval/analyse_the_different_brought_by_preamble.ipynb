```python
import json
```

```python
with open('./log_prob_temp_res/full_context/dl_e5_2.json') as f:
    without_query = json.load(f)
```

```python
with open('./log_prob_temp_res/full_context_with_query/dl_e5_2.json') as f:
    with_query = json.load(f)
```

```python
from scipy import stats

corr_lists = [[], []]

for qid in list(set(without_query.keys()) & set(with_query.keys())):
    corr_lists[0].append(without_query[qid])
    corr_lists[1].append(with_query[qid])

stats.kendalltau(corr_lists[0], corr_lists[1])
```
