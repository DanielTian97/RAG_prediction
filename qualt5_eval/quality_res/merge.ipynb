```python
import pandas as pd


for retriever in ['bm25', 'mt5', 'e5']:
    for suffix in (['']+[f'_integrated_{i}' for i in [2, 3, 5, 7]]):
        res_19 = pd.read_csv(f'{retriever}_dl_19{suffix}.csv')
        res_20 = pd.read_csv(f'{retriever}_dl_20{suffix}.csv')
        pd.concat([res_19, res_20], axis=0).to_csv(f'{retriever}_dl{suffix}.csv', index=False)
```
