```python
import pickle as pkl
import numpy as np

f = open(f'../coherence_res/dl_{_k}_{_ret}.pkl', 'rb')
matrix = pkl.load(f)
f.close()
```

```python
def cal_avg_non_diagnoal_elements(x):
    return (x.sum() - x.diagonal().sum())/(x.shape[0]*(x.shape[0]-1))

def cal_avg_upper_triangle(x):
    return (x.sum() - np.tril(x).sum())/(x.shape[0]*(x.shape[0]-1)/2)

def cal_column_avg_upper_triangle(x):
    upper_tri = (x - np.tril(x)).T
    # upper_tri = np.tril(x).T
    column_cohs = []
    for i in range(1, len(upper_tri)):
        column_cohs.append(upper_tri[i].sum()/i)
    return np.mean(column_cohs)

coh_dict = {}
for qid, mtx in matrix.items():
    # print(qid)

    _sub_cohs = []

    ## document average
    # print(doc_length_dict[qid])
    _start = 0
    for _d_start in doc_length_dict[qid][:_k]:
        if(_d_start > 1):
            submat = np.matrix(mtx)[_start:_start+_d_start,_start:_start+_d_start]
            _sub_cohs.append(cal_column_avg_upper_triangle(submat))
        _start += _d_start

    ## fixed length average
    _window = 5
    if(len(mtx) < _window):
        _window = len(mtx)
    for _start in range(0, len(mtx)-_window+1, 1):
        submat = np.matrix(mtx)[_start:_start+_window,_start:_start+_window]
        _sub_cohs.append(cal_column_avg_upper_triangle(submat))

    if(len(_sub_cohs) > 0):
        coh_dict.update({str(qid): np.mean(_sub_cohs)})
```

```python
from scipy import stats
import matplotlib.pyplot as plt

correlated_lists = [[], []]
print('should be calculated: ', len(utility_dict), 'really estimated: ', len(coh_dict))
for qid in set(utility_dict.keys()).intersection(set(coh_dict.keys())):
    correlated_lists[0].append(utility_dict[qid])
    correlated_lists[1].append(coh_dict[qid])

# print(correlated_lists[0], correlated_lists[1])
print(stats.pearsonr(correlated_lists[0], correlated_lists[1]))
print(stats.kendalltau(correlated_lists[0], correlated_lists[1]))

plt.scatter(correlated_lists[0], correlated_lists[1])
plt.show()
```
